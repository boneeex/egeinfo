```python
from itertools import permutations

# совпадения с таблицы
tab = "348 67 156 18 367 235 258 147".split(" ")
# совпадения с рисунка 
pic = "bd dg ga af fh hc cb be ed eh gf".split(" ")

print(*range(1, 9))
# перебор по буквам на рисунке
for var in permutations("abcdefgh"):
    if all(str(var.index(x) + 1) in tab[var.index(y)] for x, y in pic):
        print(" ".join(var))
```

```python
def pr_m(x, p = 2):
    for d in range(p, int(x ** 0.5) + 1):
        if x % d == 0:
            return [d] + pr_m(x // d, d)
    return [x]

# возвращает разложение на простые множители (без группировки)
print(pr_m(514))
```

```python
def sieve(n):
    lst = [*range(2, n + 1)]
    length = len(lst)
    for idx, sym in enumerate(lst):
        if sym == 0:
            continue
        if sym * sym > n:
            break
        start = sym * sym - 2
        for i in range(start, length, sym):
            lst[i] = 0
    return [i for i in lst if i]

# возвращает список простых от 2 до n
print(sieve(2547245747))
```
