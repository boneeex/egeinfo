Входные данные представлены в файле следующим образом. В следующих N   строках находятся пары натуральных чисел: номер ряда и номер места занятого кресла соответственно (первое число не превышает значения M  , а второе – K  ).

Условию задачи удовлетворяют места 1 и 2 в ряду 6: перед 1 креслом 2 занятых места, а перед 2 местом 1 занятое место.

```python
# Открываем файл с данными
f = open(r"D:\informaticsclass\egeinfo\26\26__86lva (1).txt")

# Читаем основные параметры:
# n - количество занятых мест
# m - количество рядов
# k - количество мест в каждом ряду
n, m, k = map(int, f.readline().split())

# Создаем структуру для хранения занятых мест по рядам
# row[i] будет содержать список занятых мест в ряду i+1
# Изначально добавляем m+1 как маркер конца ряда
row = [[m+1] for i in range(k)]

# Читаем информацию о занятых местах
for i in range(n):
    x, y = map(int, f.readline().split())
    row[y-1].append(x) # Добавляем занятое место в соответствующий ряд

# Инициализируем переменные для поиска оптимального решения
mx_r = 0 # Максимальный номер ряда (как можно дальше от сцены)
mx_i = 0 # Максимальный номер места

# Проходим по всем парам соседних рядов
for i in range(len(row) - 1):
    # Находим максимальное занятое место из двух соседних рядов
    # Чем больше это значение, тем дальше от сцены мы можем сесть
    if max(row[i + 1] + row[i]) >= mx_r:
        x = max(row[i + 1] + row[i])

        # Ищем подходящее место, удовлетворяющее условиям:
        # - место не должно быть занято в текущем ряду
        # - место не должно быть занято в соседнем ряду
        # - перед этим местом должно быть не более 2 занятых мест
        while (x in row[i] or x in row[i + 1]
               or len([p for p in row[i] if p < x]) > 2
               or len([p for p in row[i + 1] if p < x]) > 2):
            x -= 1 # Уменьшаем номер места, если условия не выполняются

        # Если нашли подходящее место дальше от сцены, чем предыдущие
        if x >= mx_r:
            mx_r = x # Сохраняем новый максимальный номер ряда
                                                                                                  
                                                                                                  
            mx_i = i # Сохраняем индекс ряда

# Выводим результат: номер ряда и номер места
print(mx_r, mx_i+2)
```

```python
file = open(r"D:\informaticsclass\egeinfo\26\26_17643.txt")

n = int(file.readline())
items = [list(map(int, i.split())) for i in file]
sr = sum(i[1] for i in items) / len(items)
expensive = [i for i in items if i[1] > sr]
dc = {}

for i in expensive:
    art, cost, status = i
    try:
        dc[art].append((cost, status))
    except:
        dc[art] = [(cost, status)]

res = []
for art, content in dc.items():
    res.append((
                # sold
                sum(1 for i in content if i[1] == 1),
                # remaining
                sum(1 for i in content if i[1] == 0),
                # price
                content[0][0]
                ))


res = sorted(res, key=lambda x: [-x[0], -x[2], x[1]])
print(res[0])
```
