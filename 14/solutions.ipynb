записали в системе счисления с основанием 27. Определите в 27-ричной записи числа количество цифр с чётным числовым значением, превышающим 9.

```python
from string import printable
alp = printable[:27]
def trans(n, base, alp):
    if n == 0:
        return "0"
    s = ""
    while n > 0:
        s += alp[n % base]
        n //= base
    return s[::-1]

exp = 2 * 2187 ** 567 + 729 ** 566 - 2 * 243 ** 565 + 81 ** 564 - 2 * 27 ** 563 - 6561
exp = trans(exp, 27, alp)
cnt = 0
for i in exp:
    if alp.index(i) > 9:
        cnt += 1
print(cnt)
```

№ 3775 Вариант от учеников (Уровень: Сложный)

Значение выражения 39 • 15^64 + 35^450 + 74 • 43^121 - 450035 записали в некоторой системе счисления.

Оказалось, что количество цифр 8 в этой записи максимально.

Найдите эту систему счисления и запишите её в ответе в десятичной системе счисления. Если таких несколько, укажите ту, у которой числовое значение больше.

```python
from string import printable
alp = printable
def trans(n, base, alp):
    if n == 0:
        return "0"
    s = ""
    while n > 0:
        s += alp[n % base]
        n //= base
    return s[::-1]

res = []
for base in range(9, 36):
    exp = trans(39 * 15 ** 64 + 35 ** 450 + 74 ** 121 - 450035, base, alp)
    res.append((exp.count("8"), int(exp, base), base))

print(sorted(res, reverse=True)[0][-1])
```

Значение выражения 53^123 + 65^2222 – 172^12 записали в системе счисления с основанием 7. Определите количество комбинаций цифр 6# в этой записи, где # – любая цифра от 1 до 5.

```python
from string import printable
alp = printable
def trans(n, base, alp):
    if n == 0:
        return "0"
    s = ""
    while n > 0:
        s += alp[n % base]
        n //= base
    return s[::-1]

from re import *
pattern = "6[1-5]"
pattern = f"(?=({pattern}))"
cnt = 0
for i in finditer(pattern, trans(53 ** 123 + 65 ** 2222 - 172 ** 12, 7, alp)):
    cnt += 1
print(cnt)
```
