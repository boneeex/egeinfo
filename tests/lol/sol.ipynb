```python
from functools import lru_cache

@lru_cache(None)
def f(a, m):
    if a <= 32:
        return m % 2 == 0
    
    if m == 0:
        return False
    
    h = [f(a - 2, m - 1), f(a - 3, m - 1), f(int(a / 4), m - 1)]

    if m % 2 == 1:
        return any(h)
    # внимательно смотрим на фразу (если ход не наш и m % 2 == 0)
    # при любом ходе Пети Ваня может выиграть своим первым ходом - any(h)
    # Ваня выиграл своим первым ходом после неудачного хода Пети - all(h)
    return all(h)

for s in range(33, 150):
    if f(s, 2) == True:
        print(s)
        break
```

```python
from re import *
file = open(r"D:\informaticsclass\egeinfo\tests\lol\24_25117.txt").readline()
file = file.replace("ZAHVAT", "*").replace("BROSOK", "!")
pattern = "[*][0-9A-Z]+[!]"
for i in finditer(pattern, file):
    file = file.replace(i.group(0), f"({len(i.group(0))})")

skobki = v_skobkah = mx = 0
start = 2
for end in range(2, len(file)):
    if file[end - 2] + file[end] == "()":
        skobki += 1
        v_skobkah += int(file[end - 1])
    
    while skobki > 5:
        if file[start - 2] + file[start] == "()":
            skobki -= 1
            v_skobkah -= int(file[start - 1])
        start += 1
    mx = max(mx, 60 + v_skobkah + 3 + end - start + 1)
print(mx)
```
