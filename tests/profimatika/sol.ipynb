```python
from itertools import permutations

pic = "fa ac cg ge ef ed da df db bg bc".split(" ")
tab = "357 467 145 236 1367 245 125".split(" ")

print(*range(1, 8))
for var in permutations("abcdefg"):
    if all(str(var.index(x) + 1) in tab[var.index(y)] for x, y in pic):
        print(" ".join(var))
```

```python
from itertools import *

def f(x, y, w):
    return (x <= y) and (w or (not w))

for a1, a2, a3 in product([0, 1], repeat=3):
    t = [
        (1, 1, 0),
        (a1, 1, a2),
        (1, 0, 1),
        (1, a3, 1)
    ]
    if len(t) == len(set(t)):
        for i in permutations("xyw"):
            if [f(**dict(zip(i, h))) for h in t] == [0, 0, 1, 1]:
                print(i)
```

```python
def trans(n, base):
    if n == 0:
        return "0"
    s = ""
    while n > 0:
        s += str(n % base)
        n //= base
    return s[::-1]

for n in range(10 ** 5, 1, -1):
    nw = trans(n, 3)
    n_2 = trans(nw.count("2"), 3)
    nw = nw + n_2
    n_1 = trans(nw.count("1"), 3)
    nw = nw + n_1
    n_0 = trans(nw.count("0"), 3)
    nw = nw + n_0 
    r = int(nw, 3)
    if r < 1000:
        print(n)
        break
```

```python
from turtle import *
left(90)
m = 80
tracer(0)

for i in range(36):
    forward(1 * m)
    right(36)

up()

forward(4 * m)
right(90)

down()

for i in range(8):
    forward(6 * m)
    right(90)

up()
for x in range(-30, 30):
    for y in range(-30, 30):
        goto(x * m, y * m)
        dot(2)

forward(300)
done()
```

```python
from string import printable
from itertools import product
alp = printable[:7]
cnt = 0
for var in product(alp, repeat=6):
    if var[0] != "0" and var.count("0") == 1 and sum([1 for i in var if int(i) % 2 == 0]) % 2 == 1:
        cnt += 1
print(cnt)
```
