```python
from turtle import *
speed(1000)
left(90)
m = 5000
begin_fill()
for i in range(3):
    forward(27 * m)
    right(90)
    forward(12 * m)
    right(90)
end_fill()
up()
forward(4 * m)
right(90)
forward(6 * m)
left(90)
down()
begin_fill()
for i in range(4):
    forward(83 * m)
    right(90)
    forward(77 * m)
    right(90)
end_fill()
canvas = getcanvas()
cnt = 0
for x in range(-1000, 1000):
  for y in range(-1000, 1000):
    if canvas.find_overlapping(x*m, y*m, x*m, y*m) != ():
      cnt += 1
print(cnt)
```
