Косов) Напишите программу, которая перебирает числа, большие 9 590 000, в порядке возрастания и находит первые 5 чисел, которые можно представить в виде произведения ровно семи различных простых сомножителей, среди которых ровно 2 содержат цифру 7. Запишите эти 5 чисел в порядке возрастания, справа от каждого числа запишите сумму его сомножителей, содержащих цифру 7.

```python
def pr_m(x, p = 2):
    for d in range(p, int(x ** 0.5) + 1):
        if x % d == 0:
            return [d] + pr_m(x // d, d)
    return [x]

n = 9_590_000
res = []
while len(res) < 5: 
    n += 1
    pr = pr_m(n)
    if len(set(pr)) != 7 or len(pr) != 7 or len(pr) != len(set(pr)):
        continue
    pr_7 = []
    for i in pr:
        if "7" in str(i): pr_7.append(i)
        if len(pr_7) > 2: break
    
    if len(pr_7) == 2:
        res.append((n, sum(pr_7)))
    pr_7 = 0

for i in res:
    print(i)
```

```python
from collections import Counter
from math import prod

res = []
n = 800_000
while len(res) < 6:
    n += 1
    pr = pr_m(n)
    k = 1
    for i in Counter(pr).values():
        k *= (i + 1)
    if k % 2 == 1 and prod(pr) % 2 == 1 and k > 10:
        res.append((n, k))
    k = 0

[print(*i) for i in res]
```

— символ «?» означает ровно одну произвольную цифру;

— символ «*» означает любую последовательность цифр произвольной длины; в том числе «*» может задавать и пустую последовательность.

Например, маске 123*4?5 соответствуют числа 123405 и 12300405. Среди натуральных чисел, не превышающих 10^9, найдите все числа, соответствующие маске 15*3*09 и имеющие ровно 9 делителей. В ответе запишите в первом столбце таблицы все найденные числа в порядке возрастания, а во втором столбце – соответствующие им максимальные делители, не считая самого числа

```python
from collections import Counter
from itertools import *

def pr_m(x, p=2):
    for d in range(p, int(x ** 0.5) + 1):
        if x % d == 0:
            return [d] + pr_m(x // d, d)
    return [x]

generated = []
for total_len in range(5):
    for len1 in range(total_len + 1):
        len2 = total_len - len1
        for part1 in product("0123456789", repeat=len1):
            for part2 in product("0123456789", repeat=len2):
                generated.append(int("15" + "".join(part1) + "3" + "".join(part2) + "09"))

res = []
for n in set(generated):
    k = 1
    pr = pr_m(n)
    for i in Counter(pr).values():
        k *= (i + 1)
        if k > 9:
            break
    
    if k == 9:
        res.append((n, n // min(pr)))

[print(*i) for i in res]
```

Пусть М - сумма минимального и максимального простых натуральных делителей целого числа, не считая самого числа. Если таких делителей у числа нет, то значение М считается равным нулю.
Напишите программу, которая перебирает целые числа, большие 5 400 000, в порядке возрастания и ищет среди них такие, для которых М больше 60 000 и является палиндромом, т.е. одинаково читается слева направо и справа налево. В ответе запишите в первом столбце
таблицы первые  пять найденных чисел в порядке возрастания, а во втором столбце - соответствующие им значения М.
Например, для числа 298 M = 2 + 149 = 151.

```python
def pr_m(x, p=2):
    for d in range(p, int(x ** 0.5) + 1):
        if x % d == 0:
            return [d] + pr_m(x // d, d)
    return [x]

res = []
n = 5_400_000
while len(res) < 5:
    n += 1
    pr = pr_m(n)
    if len(pr) == 1: continue
    m = min(pr) + max(pr)
    k = len(str(m))
    if len(str(m)) % 2 == 0:
        pal = str(m)[:k//2] == str(m)[k//2:][::-1]
    else:
        pal = str(m)[:k//2] == str(m)[k//2 + 1:][::-1]
    if m > 60_000 and pal:
        res.append((n, m))

[print(*i) for i in res]
```

Например, начало списка суперпростых чисел выглядит так:

1. 3 (стоит на 2 месте в списке простых чисел)
2. 5 (стоит на 3 месте в списке простых чисел)
3. 11 (стоит на 5 месте в списке простых чисел)
Среди суперпростых чисел, не превышающих 10^5, найдите все числа, соответствующие маске 1*7?7.

```python
from fnmatch import fnmatch

def prost(n):
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

pr = [i for i in range(2, 10**5 + 1) if prost(i)]
res = []

for i in range(len(pr)):
    if i + 1 in pr and fnmatch(str(pr[i]), "1*7?7"):
        res.append((pr[i], i + 1))

[print(i) for i in res]
```

найдите все числа, соответствующие маске 1234# с разными P и делящиеся на (P+2)**3. 
Если для конкретного P найдется несколько чисел, то запишите минимальное из них. 
В ответе запишите найденные числа в порядке возрастания. 
Справа от каждого числа запишите его число P.

```python
from itertools import product
from collections import defaultdict

def prost(n):
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

pr = [i for i in range(2, 10**5 + 1) if prost(i)]
res = defaultdict(list)

for i in range(0, 7):
    for var in product("1234567890", repeat=i):
        num = int("1234" + "".join(var))
        p = sum(list(map(int, list(var))))
        if p in pr and num % ((p + 2) ** 3) == 0:
            res[p] += [num]

res2 = []
for pair in res:
    res2.append(min(res[pair]))

[print(i) for i in res2]
```
