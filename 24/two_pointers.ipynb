Текстовый файл содержит строку из заглавных латинских букв и точек, всего не более 106 символов. Определите максимальное количество идущих подряд символов, среди которых не более шести точек и более 15 гласных букв (AEIOUY).

```python
from collections import defaultdict
file = open(r"D:\informaticsclass\egeinfo\24\24_M3__42nfh.txt")
file = file.readline()

d = defaultdict(int)
p = mx = start = 0
for end in range(len(file)):
    if file[end] in "AEIOUY":
        d[file[end]] += 1
    if file[end] == ".":
        p += 1

    while p > 6:
        if file[start] in "AEIOUY":
            d[file[end]] -= 1
        if file[start] == ".":
            p -= 1
        start += 1
    if p <= 6 and sum(d.values()) > 15:
        mx = max(mx, end - start + 1)
print(mx)
```

Пример:

Из строки wqe, строка wqq (можно собрать палиндром qwq) получается с помощью одной замены e -> q.

```python
from collections import Counter
file = open(r"D:\Study\egeinfo\24\24_3099.txt").readline()

k_s = Counter(file).values()
lst = [i for i in k_s if i % 2]
if len(lst) % 2 == 0:
    print(len(lst) // 2)
else:
    print((len(lst) - 1) // 2)
```
