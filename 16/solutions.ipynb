```python
from functools import lru_cache
def q(n):
    res = 0
    while n >= 21:
        res += 2
        n -= 4
    return n + 4 + res

def g(n):
    res = 0
    while n < 11240:
        res += 2
        n += 3
    return q(n) + res

@lru_cache(10)
def f(n):
    if n < 43: return g(n + 4)
    else: return 2 * f(n - 2) - f(n - 4) + 2

for i in range(2026):
    f(i)

print(f(2026))
```

```python
from functools import lru_cache

@lru_cache(None)
def f(n):
    if n < 3: return 1
    elif n > 2 and n % 2 == 0: return g(n) + f(n - 1)
    else: return f(n - 2) - 2 * g(n + 1)

@lru_cache(None)
def g(n):
    if n < 3: return 1
    elif n > 2 and n % 2 == 0: return f(n - 3) + f(n - 2)
    else: return f(n + 1) - g(n - 1)

for i in range(120):
    f(i)
    g(i)

print(g(120))
```

```python
from functools import lru_cache

@lru_cache(2)
def f(n, k):
    if k == 0:
        return 0
    elif k > 0 and n % k == 0:
        return f(n, k - 1)+ k ** 2
    else:
        return f(n, k - 1)

for i in range(123_456_789):
    f(11_223_456_789, i)
    
print(f(11_223_456_789, 123_456_789))
```
